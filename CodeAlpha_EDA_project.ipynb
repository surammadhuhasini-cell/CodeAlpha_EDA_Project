```python
import pandas as pd
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
print(df.head())
```

```python
print("Dataset shape:", df.shape)
print("Number of rows:",df.shape[0])
print("Number of columns:",df.shape[1])
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print("Average bill:",df["total_bill"].mean())
print("Average tip:",df["tip"].mean())
print("Highest bill:",df["total_bill"].max())
print("Average tip by day:")
print(df.groupby("day")["tip"].mean())
```
